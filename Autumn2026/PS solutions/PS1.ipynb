## 4. <a id='toc4_'></a>[Calculate monthly transport costs](#toc0_)

A student travels to campus on 18 days during a month and pays for two trips on each travel day. A single trip costs 24 DKK. In the following month, the fare rises by 12.5 percent and the student travels on 16 days.

1. Calculate the original monthly cost.
2. Calculate the new fare without overwriting the original fare.
3. Calculate the new monthly cost using 16 travel days.
4. Calculate the change in monthly cost.
5. Calculate what the monthly cost would have been if the fare had increased but the student had still travelled on 18 days. Store it as `price_only_cost`.
6. Print the results with clear labels.

In a short Markdown note, explain why `price_only_cost` differs from the new monthly cost.

````python
single_trip_fare = 24.0
travel_days = 18
trips_per_day = 2
fare_increase_rate = 0.125
new_travel_days = 16
````

````python
# Solution with explanation

# trip cost before the fare increase
trip_cost_before = single_trip_fare * travel_days * trips_per_day

# new trip cost after the fare increase
new_single_trip_fare = single_trip_fare * (1 + fare_increase_rate)

# new trip cost after the fare increase and reduced travel days
new_trip_cost = new_single_trip_fare * new_travel_days * trips_per_day

# new trip cost after the fare increase and reduced travel days
price_only_cost = new_single_trip_fare * travel_days * trips_per_day

# print results
# we use f-strings to format the output and include the variable values in the printed message
# with f-strings we can include expressions inside curly braces {} and they will be evaluated and included in the string

print('Original monthly travel:')
print(f'When traveling {travel_days} days, {trips_per_day} times a day with a fare price of {single_trip_fare}, the total cost is: {trip_cost_before}\n')

print('New monthly travel:')
print(f'After a fare increase of {fare_increase_rate*100}%, the new fare is: {new_single_trip_fare}')
print(f'When traveling {new_travel_days} days, {trips_per_day} times a day with the new fare price of {new_single_trip_fare}, the total cost is: {new_trip_cost}\n')

print('Price only travel:')
print(f'If we only consider the fare increase and not the reduced travel days, the total cost would be: {price_only_cost}')

# I intepret the results as follows: The agent substitutes away from spending money on travel when the fare increases by reducing the amount of days they travel.
# the agent ends up spending the same amount on travel as before the fare increase, so perhaps the agent has a budget for travel, that they are willing to spend
# always to avoid having to bike/walk as much as possible.
````

## 6. <a id='toc6_'></a>[Correct and summarize a spending series](#toc0_)

The lists below record monthly transport spending from January to June. The April value contains an extra zero.

1. Print the month and spending value for April using list indexes.
2. Calculate the average spending before correcting the error. Use `sum()` and `len()`.
3. Correct the April value in place. Do not retype the list.
4. Calculate the corrected average.
5. Use slices to calculate the average for the first three months and the last three months separately.
6. Calculate the percentage change from January to June.
7. Print the two slices and all calculated values with labels.

Add one sentence explaining why checking individual observations can matter before calculating a summary statistic.

````python
# Solution with explanation

# keep the original lists in the cell, to avoid having to re-run the previous cells if you want to re-run this cell
months = ["January", "February", "March", "April", "May", "June"]
transport_spending = [672.0, 720.0, 768.0, 7440.0, 816.0, 720.0]
print('Indexing/slicing method with lists:')

# 1.
print(f'Transport spending in {months[3]} before correction: {transport_spending[3]}')

# 2.
average_transport_spending = sum(transport_spending) / len(transport_spending)
# sum() adds all values, while len() returns the number of values in the list
print(f'Average transport spending before correcting the error: {average_transport_spending}')

# 3.
transport_spending[3] = 744.0
print(f'Transport spending in {months[3]} after correction: {transport_spending[3]}')

# 4.
average_transport_spending_corrected = sum(transport_spending) / len(transport_spending)
print(f'Average transport spending after correcting the error: {average_transport_spending_corrected}')

# 5.
average_first_three = sum(transport_spending[:3]) / len(transport_spending[:3])
average_last_three = sum(transport_spending[3:]) / len(transport_spending[3:])
print(f'Average transport spending for the first three months: {average_first_three}')
print(f'Average transport spending for the last three months: {average_last_three}')

# 6.
pct_change = transport_spending[-1]/transport_spending[0]-1
print(f'Percentage change in transport spending from January to June: {pct_change*100:.2f}%') # .2f formats the number to a float with 2 decimals
# 7. is done under each step

# EXTRA
# to easilier select the correct month a dictionary can be used instead

transport_dict = dict(zip(months, transport_spending)) # this creates a dictionary where the keys are the months and the values are the transport spending for that month
# the zip function takes two lists and returns an iterator of tuples, where the first element of each tuple is from the first list and the second element is from the second list

months = ["January", "February", "March", "April", "May", "June"]
transport_spending = [672.0, 720.0, 768.0, 7440.0, 816.0, 720.0]

# now we can easily access the transport spending for a specific month by using the month name as the key

print('\n')
print('Dictionary method:')
# e.g. 1. now becomes:
print(f'Transport spending in April: {transport_dict["April"]}')
# 2. now becomes:
average_transport_spending_dict = sum(transport_dict.values()) / len(transport_dict)
print(f'Average transport spending: {average_transport_spending_dict}')
# 3. now becomes:
transport_dict["April"] = 744.0
print(f'Transport spending in April after correction: {transport_dict["April"]}')
# 4. now becomes:
average_transport_spending_corrected_dict = sum(transport_dict.values()) / len(transport_dict)
print(f'Average transport spending after correction: {average_transport_spending_corrected_dict}')
# 5. now becomes:
average_first_three_dict = sum(list(transport_dict.values())[:3]) / len(list(transport_dict.values())[:3])
average_last_three_dict = sum(list(transport_dict.values())[3:]) / len(list(transport_dict.values())[3:])
print(f'Average transport spending for the first three months: {average_first_three_dict}')
print(f'Average transport spending for the last three months: {average_last_three_dict}')
# 6. now becomes:
pct_change_dict = transport_dict["June"]/transport_dict["January"]-1
print(f'Percentage change in transport spending from January to June: {pct_change_dict*100:.2f}%')

# some things are easier with lists like indexing and slicing, while others are easier with dictionaries like accessing values by key.
````

## 7. <a id='toc7_'></a>[Find and fix the errors](#toc0_)

The code below is meant to calculate gross earnings, the tax payment, net earnings and the net wage per hour, using the same figures as Task 5. It contains four separate problems.

Copy it into the empty code cell and run it. Only the first problem announces itself with an error message. Once you have fixed that one, the code runs all the way to the end without complaining — and still prints wrong numbers. Running without an error is not evidence that a calculation is right.

Fix one problem at a time, rerun after each fix, and check every printed value against the number you expect.

```python
hourly_wage = 145.0
hours_worked = "84"
tax_rate = 0.36

gross_earnings = hourly_wage + hours_worked
tax_payment = gross_earnings * tax_rate
net_earnings = tax_payment
net_hourly_wage = hourly_wage * 1 - tax_rate

print("Gross earnings:", gross_earnings, "DKK")
print("Tax payment:", tax_payment, "DKK")
print("Net earnings:", net_earnings, "DKK")
```

Write one sentence for each problem you corrected.

**Problems corrected:**

1. `TypeError`: `hours_worked` was a string instead of a number. Changed it to a float.
2. Multiplication v. addition: `gross_earnings` should be calculated with `*` instead of `+`.
3. Missing variable: `net_earnings` should be calculated as `gross_earnings - tax_payment` instead of just `tax_payment`.
4. Order of operations: `net_hourly_wage` should be calculated as `hourly_wage * (1 - tax_rate)` instead of `hourly_wage * 1 - tax_rate`.

The corrected code is found below.

````python
hourly_wage = 145.0
hours_worked = 84.0
tax_rate = 0.36

gross_earnings = hourly_wage * hours_worked
tax_payment = gross_earnings * tax_rate
net_earnings = gross_earnings - tax_payment
net_hourly_wage = hourly_wage * (1 - tax_rate)

print("Gross earnings:", gross_earnings, "DKK")
print("Tax payment:", tax_payment, "DKK")
print("Net earnings:", net_earnings, "DKK")
print("Net wage per hour:", net_hourly_wage, "DKK")
````
